<h2 style="color: #1E3A8A; font-size: 24px; border-bottom: 2px solid #eaeaea; padding-bottom: 5px; margin-top: 30px;">
    1. Caricamento dei Dataset e Gestione delle Risorse
</h2>
L'importazione dei dataset è stata progettata per minimizzare il consumo di RAM e velocizzare i tempi di caricamento dei dataset. Sono state ottimizzate **3 caratteristiche principali**:

#### 1. Data Downcasting (Riduzione dei tipi di dato)
Di default, Pandas assegna formati a 64 bit molto pesanti. Per ridurre l'impronta in memoria si è applicato un mapping esplicito (`dtypes`) prima della lettura:
* **`int32` e `int16`:** Applicati a ID e contatori spaziali/temporali (`timestamp`, `txId`, `addressId`, `position`) per dimezzare o ridurre a un quarto lo spazio occupato, senza rischio di overflow.
* **`float32` vs `float64`:** Le tariffe (`fee`) e i puntatori (`prevTxId`, `prevTxpos`) sono stati ridotti a 32 bit (necessari per supportare i valori nulli degli input *Coinbase*). Solo l'attributo `amount` è stato mantenuto a 64 bit (`float64`) per evitare qualsiasi perdita di precisione sui decimali dei Satoshi.
* **`category`:** Utilizzato per `isCoinbase`, riduce drasticamente il peso della variabile mappando internamente i valori univoci rispetto all'uso di stringhe o interi classici.

#### 2. Scelta dei Motori di Parsing
* **PyArrow:** I file principali sono stati letti con `engine='pyarrow'` (in C++), sfruttando il multi-threading per abbattere i tempi di caricamento di file di grandi dimensioni.
* **Python Engine:** Il file `satoshiDiceInfos.tsv` presenta separatori irregolari. Per parsarlo correttamente è stata utilizzata l'espressione regolare `r'\t+|\s{2,}'`.

#### 3. Feature Selection a Tempo Zero
Immediatamente dopo l'importazione, la colonna `scripttype` è stata rimossa (`outputs.drop`). Essendo irrilevante ai fini della costruzione del grafo, scartarla subito permette di liberare prezioso spazio in RAM per le successive operazioni di *join*.

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sea
import matplotlib.dates as mdates
import matplotlib.ticker as ticker 
import networkx as nx

sea.set_theme(style='whitegrid')
```

```python
# In questo modo vado a ridurre la dimensione effettiva di ogni tabella caricata in ram

dtypes_transactions = {
    'timestamp': 'int32',
    'blockId': 'int32',
    'txId': 'int32',
    'isCoinbase': 'category',
    'fee': 'float32'
}

dtypes_inputs = {
    'txId': 'int32',
    'prevTxId': 'float32',
    'prevTxpos': 'float32'
}

dtypes_outputs = {
    'txId': 'int32',
    'position': 'int16',
    'addressId': 'int32',
    'amount': 'float64'
}

dtypes_mapping = {
    'hash': 'object',
    'addressId': 'int32'
}
```

```python
#Nomi delle colonne delle varie tabelle
cols_transactions = ['timestamp','blockId','txId','isCoinbase','fee']
cols_inputs = ['txId', 'prevTxId', 'prevTxpos']
cols_outputs = ['txId', 'position', 'addressId', 'amount', 'scripttype']
cols_mapping = ['hash', 'addressId']

#Caricamento delle tabelle in ram
transactions = pd.read_csv('DataSets/transactions.csv',header=None,names=cols_transactions, dtype=dtypes_transactions, engine='pyarrow')
inputs = pd.read_csv('DataSets/inputs.csv', header=None, names= cols_inputs,dtype=dtypes_inputs,engine='pyarrow')
outputs = pd.read_csv('DataSets/outputs.csv', header=None,names=cols_outputs,dtype=dtypes_outputs, engine='pyarrow')
mapping = pd.read_csv('DataSets/mapAddr2Ids8708820.csv', header=None,names=cols_mapping,dtype=dtypes_mapping, engine='pyarrow')
satoshi_addresses=pd.read_csv('DataSets/satoshiDiceInfos.tsv',sep=r'\t+|\s{2,}',skiprows=1,engine='python')

outputs= outputs.drop(columns=['scripttype']) #scripttype non serve per il progetto 
```

<h2 style="color: #1E3A8A; font-size: 24px; border-bottom: 2px solid #eaeaea; padding-bottom: 5px; margin-top: 30px;">
    2. Analisi dell'Incidenza delle Transazioni di Bet sul Traffico Globale (Scelte Tecniche)
</h2>

Il primo obiettivo dell'analisi generale richiede di calcolare e contestualizzare l'impatto di Satoshi Dice sull'intera rete Bitcoin, misurando la percentuale di transazioni di *bet* rispetto al volume complessivo del traffico *on-chain*, osservato sotto diverse granularità temporali. 

Per rispondere a questa richiesta, la preparazione dei dati e l'elaborazione analitica sono state suddivise nei seguenti passaggi logici:

#### 1. Identificazione e Flagging delle Bet
Poiché Satoshi Dice non richiede registrazione, ogni scommessa è identificabile esclusivamente tramite l'indirizzo di destinazione. Sfruttando il file di mapping, sono stati filtrati e isolati gli `addressId` associati ai *vanity address* del servizio (riconoscibili dal prefisso `1dice`). Successivamente, incrociando questi ID con la tabella degli `outputs`, è stato possibile estrarre gli identificatori univoci (`txId`) delle transazioni contenenti almeno una puntata diretta al servizio. Infine, la tabella principale delle transazioni è stata arricchita con una variabile booleana (`is_bet`), fondamentale per le successive aggregazioni.

#### 2. Pre-processing Temporale
I `timestamp` nativi dei blocchi, originariamente espressi nel tempo UNIX, sono stati convertiti in formato `datetime` per permettere l'indicizzazione temporale del dataset. L'analisi è stata focalizzata sull'anno **2012**; tale scelta è coerente con il ciclo di vita del servizio, il cui lancio ufficiale è avvenuto in tale anno, risultando in un'incidenza nulla del traffico nei periodi precedenti (inclusi nel dataset a partire dal blocco genesis del 2009).

#### 3. Calcolo delle Metriche e Aggregazione (Resampling)
Per valutare l'andamento nel tempo, la serie storica è stata ricampionata (tramite la funzione `resample` di Pandas) calcolando aggregazioni su tre diverse finestre temporali:
* **Giornaliera (D)**
* **Settimanale (W)**
* **Mensile (ME)**

Per ciascun periodo, è stata calcolata l'incidenza percentuale matematica:  
$\text{Percentuale Bet} = \left( \frac{\text{Somma delle Transazioni di Bet}}{\text{Conteggio Totale delle Transazioni}} \right) \times 100$

#### 4. Smoothing delle Serie Storiche (Rolling Window)
Dato che il traffico giornaliero su blockchain presenta una forte volatilità, per i dati a granularità giornaliera è stata applicata una **media mobile a 14 giorni (Rolling Window)**. Tale scelta  è finalizzata a filtrare il rumore stocastico e le fluttuazioni, permettendo di isolare il trend di crescita effettivo e rendendo la curva di tendenza più leggibile rispetto al dato grezzo.

#### 5. Scelte di Visualizzazione
Per la rappresentazione visiva si è optato per una griglia di tre grafici a linee (*lineplot*) sovrapposti e con asse Y condiviso. Questa configurazione permette di:
* Valutare istantaneamente il macro-trend mensile, evidenziando la rapidissima adozione del servizio nel corso del 2012.
* Analizzare la volatilità a breve termine tramite la granularità settimanale e giornaliera.
* Notare l'entità dei volumi grazie all'area colorata (*fill_between*) che restituisce un immediato impatto visivo del "peso" di Satoshi Dice sulla blockchain di Bitcoin.

```python
def dataframes_elaboration(mapping, satoshi_addresses, transactions, outputs):
    
    tx = transactions.copy()

    #ottengo una series filtrando mapping, tenendo solo le righe che hanno come addressId un indirizzo satoshidice
    satoshi_id= mapping[mapping['hash'].isin(satoshi_addresses['Address'])]['addressId']
    
    # Prendo tutti gli Id degli output che sono delle bet
    bet_outputs = outputs[outputs['addressId'].isin(satoshi_id)]
    bet_ids=bet_outputs['txId'].unique()

    # aggiungo una colonna con true/false per definire se una transazione e' una bet
    tx['is_bet'] =  tx['txId'].isin(bet_ids)
    
    # Converto i timestamp in dateTime (in secondi)
    tx['datetime'] =pd.to_datetime(tx['timestamp'], unit='s')
    
    # imposto la colonna datetime come indice della tabella
    tx = tx.set_index('datetime')

    # filtro per l'anno 2012
    tx_2012 = tx[tx.index.year ==2012]
    
    return tx_2012, tx,satoshi_id,bet_outputs

    
def calculate_draw(dati,code_period,name_period,ax,graph_color):
    """Funzione che calcola le percentuali e mostra i grafici in base al periodo scelto (giorno,mese,anno)"""

    # resample del periodo
    stats = dati.resample(code_period).agg(
        total_tx = ('txId','count'),
        bet_tx = ('is_bet', 'sum')
    )
    
    # calcolo della percentuale 
    stats['bet_percentage'] = (stats['bet_tx'] / stats['total_tx'])  * 100

    # reset dell'indice per estrarre la data come indice colonna per il grafico
    stats = stats.reset_index()

    # disegno del grafico
    sea.lineplot(data=stats,x='datetime', y='bet_percentage', ax=ax, color=graph_color, linewidth=2)
    ax.fill_between(stats['datetime'], stats['bet_percentage'], alpha=0.2, color=graph_color)

    # media mobile (solo per il grafico giornaliero)
    if code_period == 'D':
        # calcolo rolling window a 7 giorni
        stats['rolling_14d'] = stats['bet_percentage'].rolling(window=14).mean()
        
        # disegno la media mobile
        sea.lineplot(data=stats, x='datetime', y='rolling_14d', ax=ax, color='#F23645', linewidth=2.5, label='Media Mobile (14 gg)')
        # posiziono la legenda
        ax.legend(loc='upper left', fontsize=11)
    
    # label del grafico 
    ax.set_title(f'Andamento {name_period}', fontsize=14)
    ax.set_xlabel('')
    ax.set_ylabel('Bet (%)',fontsize=12)

    #mette i ticker in corrispodenza di ogni mese
    ax.xaxis.set_major_locator(mdates.MonthLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))

    # Titolo e Labels
    ax.set_title(f'Incidenza Percentuale delle Bet sul Traffico Globale ({name_period})', fontsize=14, fontweight='bold')
    ax.set_xlabel('Timeline', fontsize=12)
    ax.set_ylabel('Incidenza Bet (%)', fontsize=12)
    
    # griglia tratteggiata e sfumata
    ax.grid(True, linestyle='--',alpha=0.5)
    
    # aggiunta del % all'asse y
    ax.yaxis.set_major_formatter(ticker.PercentFormatter(decimals=1))
    
```

```python

# elaborazione dei dataframes
tx_2012, tx,satoshi_id,bet_outputs= dataframes_elaboration(mapping, satoshi_addresses, transactions, outputs)

# Figura che contiene 3 grafici (uno per ogni riga)
fig, axes = plt.subplots(3, 1,figsize=(14, 12), sharey=True)
fig.suptitle('Percentuale di Transazioni di Bet nel 2012', fontsize=18,fontweight='bold')

# Stampa dei tre grafici
calculate_draw(tx_2012,'D','Giornaliero', axes[0],'#089981')
calculate_draw(tx_2012, 'W','Settimanale', axes[1],'#3B82F6')
calculate_draw(tx_2012,'ME','Mensile',axes[2],'#F7931A')

plt.tight_layout()
plt.subplots_adjust(top=0.90) # Aggiustamento per il titolo
plt.show()


```

L'esame visivo delle serie storiche restituisce una chiara fotografia dell'impatto esplosivo di Satoshi Dice sull'ecosistema Bitcoin nel corso del 2012. <h2 style="color: #1E3A8A; font-size: 24px; border-bottom: 2px solid #eaeaea; padding-bottom: 5px; margin-top: 30px;">
    3. Popolarità degli Indirizzi di Betting (Scelte Tecniche)
</h2>

Il secondo obiettivo dell'analisi generale richiede di misurare il successo e la popolarità dei vari indirizzi (identificati dal prefisso `1dice`) messi a disposizione da Satoshi Dice. Per fornire una panoramica completa, la popolarità è stata misurata seguendo due metriche distinte:

1. **Frequenza di utilizzo:** il numero totale di scommesse ricevute da ciascun indirizzo.
2. **Volume monetario:** l'ammontare complessivo (`amount`) trasferito verso ciascun indirizzo.

Per raggiungere questo scopo e preparare i dati alla visualizzazione, il codice è stato strutturato nei seguenti passaggi:

* **Aggregazione dei Dati:** partendo dal dataframe `bet_outputs` (precedentemente filtrato), le transazioni sono state raggruppate per `addressId`. tramite la funzione `agg()`, sono state estratte simultaneamente le due metriche richieste: il conteggio delle scommesse (`num_bet`) e la somma degli importi giocati (`tot_amount`).
* **Mapping e denominazione:** l'ID numerico dell'indirizzo è scarsamente comunicativo. Pertanto, la tabella aggregata è stata unita prima al dataframe di `mapping` per recuperare l'hash originale dell'indirizzo, e successivamente al file contenente i metadati di Satoshi Dice. Questo ha permesso di associare a ogni barra del grafico la rispettiva etichetta `Name` (es. *"lessthan 64000"*, che descrive la condizione di vincita del gioco), rendendo la lettura dei risultati immediata.
* **Conversione di Valuta:** l'importo aggregato (`tot_amount`) , nativamente espresso in Satoshi, genererebbe numeri interi con troppe cifre, rendendo l'asse X illeggibile. Prima della visualizzazione, il volume totale è stato diviso per $10^8$, convertendolo formalmente in Bitcoin (BTC).
* **Scelte di Visualizzazione:** si è optato per due **grafici a barre orizzontali** (bar chart) affiancati. La scelta dell'orientamento orizzontale è ottimale quando si gestiscono categorie testuali relativamente lunghe (come i nomi delle bet), in quanto evita l'accavallamento delle etichette sull'asse X. Inoltre, l'ordinamento forzato facilita il confronto visivo tra le due metriche.

```python

def calculate_address_popularity(bet_outputs, mapping, satoshi_addresses):
    """
    Raggruppa le scommesse per indirizzo ,calcola volumi e bet, e unisce i dati con i nomi reali di satoshidice."""
    #   raggruppo per indirizzo e calcolo num_bet e tot_amount
    bet_info = bet_outputs.groupby('addressId' ).agg(num_bet=( 'txId', 'count'),tot_amount=('amount', 'sum')).reset_index()

    #unisco bet_info  con mapping per avere l'indirizzo hash
    bet_info_hash = bet_info.merge( mapping, on='addressId')

    # merge tra bet_info e satoshi_addresses per avere l'etichetta 'Name'
    bet_info_final = bet_info_hash.merge(satoshi_addresses[['Address', 'Name']], left_on='hash', right_on='Address')
    
    return bet_info_final

def plot_popularity_by_num_bets(bet_info_final, ax,color_num_bets):
    """
    Disegna il grafico a barre per la popolarità basata sul numero di bet all'interno dell'asse ax specificato.
    e ritorna il dataframes di bet ordinate in base al numero"""
    #ordino i valori, per la rappresentazione grafica
    sorted_num_bets = bet_info_final.sort_values('num_bet', ascending=True)

    #configurazione del grafico a barre
    ax.barh(sorted_num_bets['Name'], sorted_num_bets['num_bet'], color=color_num_bets, alpha=0.8, edgecolor='black',linewidth=0.4)
    ax.set_title('Popolarità in base al numero di scommesse', fontsize=14, fontweight='bold')
    ax.set_xlabel('Numero totale di scommesse', fontsize=12)
    ax.grid(axis='x',linestyle='--', alpha=0.5) 
    
    ax.xaxis.set_major_formatter(ticker.FuncFormatter(lambda x, p: format(int(x),',')))
    return sorted_num_bets


def plot_popularity_by_total_amount(bet_info_final, ax,color_total_amount):
    """Disegna il grafico a barre per la popolarità basata sul  volume totale all'interno dell'asse ax specificato."""
    #ordino i valori, per la rappresentazione grafica
    sorted_total_amount = bet_info_final.sort_values('tot_amount', ascending=True)
    
    #aggiungo una colonna con la conversione da satoshi a BTC
    sorted_total_amount['tot_amount_btc'] = sorted_total_amount['tot_amount'] / 100000000

    ax.barh(sorted_total_amount['Name'], sorted_total_amount['tot_amount_btc'], color=color_total_amount, alpha=0.8, edgecolor='black',linewidth=0.4)
    ax.set_title('Popolarità per Volume Scommesso', fontsize=14, fontweight='bold')
    ax.set_xlabel('Volume Scommesso (BTC)',fontsize=12)
    ax.grid(axis='x', linestyle='--', alpha=0.5)
    
    ax.xaxis.set_major_formatter(ticker.FuncFormatter(lambda x, p: format(int(x), ',')))


```

```python
# calcolo i dati di popolarità 
bet_info_final = calculate_address_popularity(bet_outputs, mapping, satoshi_addresses)

# figura globale(1 riga,2 colonne)
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

#  disegno il primo grafico a barre  
sorted_num_bets = plot_popularity_by_num_bets(bet_info_final, axes[0],'#0ECB81')

#  disegno il secondo grafico a barre 
plot_popularity_by_total_amount(bet_info_final, axes[1],'#F23645')

plt.tight_layout()
plt.show()
```

<h2 style="color: #1E3A8A; font-size: 24px; border-bottom: 2px solid #eaeaea; padding-bottom: 5px; margin-top: 30px;">
    4. Distanza tra Transazioni di Bet e Payout (Scelte Tecniche)
</h2>

Il terzo obiettivo dell'analisi generale mira a valutare la latenza del servizio Satoshi Dice, misurando la distanza in termini di tempo e di blocchi tra il momento in cui un utente invia una scommessa (bet) e il momento in cui il sistema on-chain eroga il pagamento.

Per ricostruire questo legame logico, il codice si appoggia all'architettura a UTXO (Unspent Transaction Output) di Bitcoin. I passaggi analitici sono i seguenti:

**Collegamento Logico (UTXO Matching):** Una transazione di payout è considerata associata a una transazione di bet se un input della transazione di payout consuma un UTXO (output non speso) generato dalla transazione di bet originale. Operativamente, è stata effettuata una *inner join* tra la tabella `bet_outputs` e la tabella `inputs`. Il matching si basa su una chiave composta: l'identificatore della transazione (`txId`) e la posizione dell'output (`position`) devono coincidere esattamente con la transazione precedente (`prevTxId`) e la posizione di origine (`prevTxpos`) dichiarate nell'input speso.
**Recupero delle Metriche Temporali e Spaziali:** Una volta accoppiate la transazione di origine e quella di destinazione, il dataframe risultante è stato incrociato due volte con la tabella principale `transactions` per estrarre l'altezza del blocco (`blockId`) e il tempo di mining (`timestamp`) di entrambe le operazioni.
* **Calcolo della Distanza:** Sono state calcolate due metriche differenziali per ogni singola coppia bet-payout:
    1. **Distanza in blocchi:** calcolata come differenza matematica tra `block_payout` e `block_bet`.
    2. **Distanza temporale:** calcolata come differenza tra i timestamp e divisa per 60 per esprimerla in minuti.
* **Scelte di Visualizzazione e Filtraggio Outlier:** Essendo il processo di payout automatizzato, la distribuzione dei dati presenta una fortissima asimmetria, con la stragrande maggioranza dei valori concentrata verso lo zero. Per questo motivo, entrambi i grafici adottano una **scala logaritmica sull'asse Y** (fondamentale per non schiacciare visivamente i valori meno frequenti). Per esplorare a fondo il dataset e analizzare anche la "coda lunga" delle operazioni ritardate, i grafici mostrano una finestra visiva allargata fino a 60 blocchi di distanza e 10 ore (600 minuti) di attesa.

```python
def calculate_bet_payout_distances(bet_outputs, inputs, transactions):
    """Calcola la distanza in tempo e in blocchi tra una scommessa (bet) e il suo rispettivo pagamento (payout).
    """
    # associo ad ogni scommessa il rispettivo pagamento usando le chiavi composte
    bet_inputs = bet_outputs.merge(inputs,left_on=['txId', 'position'],right_on=['prevTxId', 'prevTxpos'],suffixes=('_bet', '_payout'))
    
    # prendo soltanto gli id delle scommesse e dei pagamenti
    bet_inputs = bet_inputs[['txId_bet', 'txId_payout']]

    #associo ad ogni scommessa il tempo e il suo blocco
    tx_bet = bet_inputs.merge(transactions, left_on='txId_bet', right_on='txId')
    #rinomino le colonne (per differenziale da quella della successiva join)
    tx_bet = tx_bet.rename(columns={'timestamp': 'time_bet', 'blockId': 'block_bet'})
    tx_bet = tx_bet[['txId_bet', 'txId_payout', 'time_bet', 'block_bet']] 

    # recupero per ogni pagamento il tempo e il suo blocco
    bet_payout = tx_bet.merge(transactions, left_on='txId_payout', right_on='txId')
    bet_payout = bet_payout.rename(columns={'timestamp': 'time_payout', 'blockId': 'block_payout'})
    
    # calcolo  delle distanze
    bet_payout['time_distance'] = bet_payout['time_payout'] - bet_payout['time_bet']
    bet_payout['block_distance'] = bet_payout['block_payout'] - bet_payout['block_bet']
    bet_payout['time_dist_min'] = bet_payout['time_distance'] / 60 # tempo in minuti
    
    #pulisco il dataframe mantenendo solo ciò che mi serve
    final_columns = ['txId_bet', 'time_bet', 'block_bet', 'txId_payout', 'time_payout', 'block_payout', 'time_distance', 'block_distance', 'time_dist_min']
    
    return bet_payout[final_columns]


def plot_block_distance_distribution(bet_payout_data, ax):
    """Disegna il countplot della distanza in blocchi tra bet e payoutall'interno dell'asse ax, rimuovendo gli outlier oltre i 12 blocchi."""
    
    # filtro gli outlier
    without_outliers = bet_payout_data[bet_payout_data['block_distance'] <= 60]

    #config del countploit
    sea.countplot(data=without_outliers, x='block_distance', hue='block_distance', palette='crest_r', legend=False,ax=ax)

    # config estetica
    ax.set_yscale('log')
    ax.set_title('Distanza in Blocchi tra Bet e Payout (fino a 60 blocchi)', fontsize=14, fontweight='bold')
    ax.set_xlabel('Numero di Blocchi di Distanza', fontsize=12)
    ax.set_ylabel('Frequenza (Scala Logaritmica)', fontsize=12)
    ax.grid(axis='y', linestyle='--', alpha=0.5)


def plot_time_distance_distribution(bet_payout_data, ax,hist_color):
    """
    Disegna l'istogramma della distanza temporale in minuti tra bet e payout all'interno dell'asse ax, limitando l'analisi alle prime 2 ore.
    """
    # filtro gli outlier (mostro i dati tra 0 e 120 minuti)
    time_without_outliers = bet_payout_data[(bet_payout_data['time_dist_min'] >= 0) & (bet_payout_data['time_dist_min'] <= 600)]

    # configurazione istogramma
    sea.histplot(data=time_without_outliers, x='time_dist_min', bins=60,color=hist_color,ax=ax ,edgecolor='black', linewidth=0.25 )

    # tick sull'asse X ogni 25 minuti
    ax.set_xticks(range(0, 601, 25))
    ax.set_xticklabels(range(0, 601, 25))
    
    # configurazione estetica
    ax.set_yscale('log')
    ax.set_title('Distanza Temporale tra Bet e Payout (entro 10h)', fontsize=14, fontweight='bold')
    ax.set_xlabel('Distanza Temporale (Minuti)', fontsize=12)
    ax.set_ylabel('Frequenza (Scala Logaritmica)', fontsize=12)

    ax.grid(axis='y', linestyle='--', alpha=0.5)
```

```python
# dataframes delle transazioni con aggiunta le nuove colonne distanza blocchi e tempo
distance_data = calculate_bet_payout_distances(bet_outputs, inputs, tx)

# figura (2 righe, 1 colonna)
fig, axes = plt.subplots(2, 1, figsize=(14, 12))

# funzioni che disegnano i rispettivi grafici
plot_block_distance_distribution(distance_data, axes[0])
plot_time_distance_distribution(distance_data, axes[1], '#FF9800')

plt.tight_layout()
plt.show()
```

L'analisi visiva delle distribuzioni spaziali (distanza in blocchi) e temporali (minuti) evidenzia l'efficienza di Satoshi Dice. Entrambi i grafici mostrano un netto decadimento esponenziale, da cui si possono trarre tre conclusioni fondamentali sul funzionamento del servizio:

1. **Reattività Istantanea (Distanza Temporale):** L'istogramma temporale presenta un picco enorme in corrispondenza dei primissimi minuti (0-2 min). <h2 style="color: #1E3A8A; font-size: 24px; border-bottom: 2px solid #eaeaea; padding-bottom: 5px; margin-top: 30px;">
    5. Analisi Dettagliata sui Top 3 Indirizzi: Distribuzione Temporale (Scelte Tecniche)
</h2>

Il quarto obiettivo del progetto richiede di restringere l'analisi ai soli 3 indirizzi risultati più popolari nel punto precedente, al fine di esaminare i pattern comportamentali dei giocatori e studiare come le loro puntate si distribuiscono nel corso del tempo.

Per preparare i dati a questa analisi multi-livello, il codice è stato organizzato nei seguenti step logici:

* **Filtraggio e Arricchimento del Dato:** Tramite l'operazione `.tail(3)` applicata al dataframe ordinato per popolarità, sono stati isolati gli ID dei tre indirizzi dominanti. Una sequenza di operazioni di `merge` (inner e left join) ha permesso di incrociare questi ID con le tabelle delle scommesse e delle transazioni, recuperando l'esatto momento di esecuzione (`datetime`), le commissioni pagate (`fee`) e l'etichetta testuale della giocata (`Name`).
* **Rifattorizzazione temporale:** Il timestamp assoluto non è adatto per studiare la stagionalità. Pertanto, utilizzando le proprietà dell'accessore `.dt` di Pandas, la marca temporale è stata scomposta in tre nuove variabili cicliche e categoriche: l'ora del giorno (`hour`), il giorno della settimana (`day_name`) e il mese dell'anno (`month`).
* **Scelte di Visualizzazione:** Si è optato per una dashboard verticale composta da tre countplot a barre raggruppate (`countplot` con parametro `hue='Name'`). Questa impostazione grafica è ideale per confrontare simultaneamente i volumi dei tre indirizzi all'interno del medesimo intervallo temporale 
* **Gestione degli Assi e degli Spazi Vuoti:** Nei grafici giornaliero e mensile è stato forzato tramite il parametro `order` un ordinamento cronologico assoluto (da Lunedì a Domenica e da Gennaio a Dicembre). In questo modo, l'asse X rispetta la naturale percezione del tempo umano, costringendo il grafico a mostrare anche i mesi in cui il volume di gioco era pari a zero (Gennaio - Aprile), evitando che Pandas ordinasse le barre casualmente .

```python
def get_top3_temporal_data(sorted_num_bets, bet_outputs, tx, bet_info_final):
    """Filtra i dati per i top 3 indirizzi più popolari, esegue i merge per recuperare le date e i nomi, ed estrae le feature temporali (ora, giorno, settimana). """
    # isolo i top 3 address
    top3_address_id = sorted_num_bets.tail(3)[['addressId']]
    
    # merge in modo da ottene i migliori 3 outputs
    top3_bets_outputs = bet_outputs.merge(top3_address_id, on='addressId', how='inner')
    
    tx_reset = tx.reset_index() # resetto l'index per avere datetime come colonna

    # merge con tx in modo da prendere tutte le informazioni 
    top3_full = top3_bets_outputs.merge(tx_reset[['datetime', 'txId', 'fee']], on='txId', how='inner')

    # aggiungo  il nome "less then numero" della bet mappandolo tramite l'addressId
    top3_full = top3_full.merge(bet_info_final[['addressId', 'Name']], on='addressId', how='left')

    # creazione delle colonne che rappresentano il tempo in formati diversi 
    top3_full['hour'] = top3_full['datetime'].dt.hour
    top3_full['day_name'] = top3_full['datetime'].dt.day_name()
    top3_full['month'] = top3_full['datetime'].dt.month
    
    return top3_full


def plot_hourly_distribution(data, ax, custom_palette):
    """ Disegna la distribuzione delle scommesse per ora del giorno (0-23). """
    # configurazione il countplot
    sea.countplot(data=data, x='hour', hue='Name', palette=custom_palette, ax=ax)
    
    # formattazione estetica di titoli e assi
    ax.set_title('Distribuzione per Ora del Giorno (UTC)', fontsize=14, fontweight='bold')
    ax.set_xlabel('Ora (0-23)', fontsize=12)
    ax.set_ylabel('Numero di Bet', fontsize=12)
    
    # setting griglia e legenda per migliorare la leggibilità
    ax.legend(title='Tipo di Giocata')
    ax.grid(axis='y', linestyle='--', alpha=0.5)


def plot_daily_distribution(data, ax, custom_palette):
    """ Disegna la distribuzione delle scommesse per giorno della settimana."""
    
    # scpecifico l'ordine delle colonne
    days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

    # configurazione del countplot
    sea.countplot(data=data, x='day_name', hue='Name', palette=custom_palette, order=days_order, ax=ax)
    
    # formattazione
    ax.set_title('Distribuzione per Giorno della Settimana', fontsize=14, fontweight='bold')
    ax.set_xlabel('Giorno', fontsize=12)
    ax.set_ylabel('Numero di Bet', fontsize=12)
    
    # etichette dell'asse x in ita per la visualizzazione
    etichette_ita = ['Lunedì', 'Martedì', 'Mercoledì', 'Giovedì', 'Venerdì', 'Sabato', 'Domenica']
    ax.set_xticks(range(len(etichette_ita))) # Fissa il numero di ticks per evitare il warning
    ax.set_xticklabels(etichette_ita)
    
    ax.legend(title='Tipo di Giocata')
    ax.grid(axis='y', linestyle='--', alpha=0.5)


def plot_monthly_distribution(data, ax, custom_palette):
    """Disegna la distribuzione delle scommesse per mese dell'anno (1-12)."""
    # forzo i 12 mesi per mostrare anche i mesi vuoti (ad esempio: Febbraio e Marzo)
    tutti_i_mesi = range(1, 13)

    #configuro il countplot
    sea.countplot(data=data, x='month', hue='Name', palette=custom_palette, order=tutti_i_mesi, ax=ax)

    # formattazione
    ax.set_title("Distribuzione per Mese dell'Anno 2012", fontsize=14, fontweight='bold')
    ax.set_xlabel('Mese', fontsize=12)
    ax.set_ylabel('Numero di Bet', fontsize=12)

    #etichette dell'asse x dei mesi
    mesi_ita = ['Gennaio', 'Febbraio', 'Marzo', 'Aprile', 'Maggio', 'Giugno', 'Luglio', 'Agosto', 'Settembre', 'Ottobre', 'Novembre', 'Dicembre']
    ax.set_xticks(range(len(mesi_ita))) 
    
    # Rotazione di 45 gradi e allineamento a destra per evitare sovrapposizioni visive
    ax.set_xticklabels(mesi_ita, rotation=45, ha='right', fontsize=10) 
    
    ax.legend(title='Tipo di Giocata')
    ax.grid(axis='y', linestyle='--', alpha=0.5)
```

```python
# prepara il dataset
top3_temporal_data = get_top3_temporal_data(sorted_num_bets, bet_outputs, tx, bet_info_final)

project_palette = ['#F23645', '#1E3A8A', '#0ECB81']

# inizializza la figura con 3 righe
fig, axes = plt.subplots(3, 1, figsize=(14, 18))
fig.suptitle('Distribuzione Temporale delle Bet (Top 3 Indirizzi)', fontsize=20, fontweight='bold')

#disegna i grafici passando i dati, gli assi e i colori
plot_hourly_distribution(top3_temporal_data, axes[0], project_palette)
plot_daily_distribution(top3_temporal_data, axes[1], project_palette)
plot_monthly_distribution(top3_temporal_data, axes[2], project_palette)

plt.tight_layout()
plt.subplots_adjust(top=0.93) # lascia spazio al titolo
plt.show()
```

<h2 style="color: #1E3A8A; font-size: 24px; border-bottom: 2px solid #eaeaea; padding-bottom: 5px; margin-top: 30px;">
    6. Analisi di Correlazione tra Fee e Amount Scommesso (Scelte Tecniche)
</h2>

Il prossimo passo dell'analisi sui tre indirizzi più popolari consiste nel verificare l'eventuale presenza di una correlazione  tra l'importo scommesso dall'utente (`amount`) e la commissione di transazione volontaria pagata ai miner (`fee`). L'obiettivo è capire se i giocatori ad alti volumi tendessero a pagare commissioni più elevate per garantirsi una priorità nell'elaborazione della scommessa.

Per condurre questa indagine statistica e visiva, l'approccio è stato il seguente:

* **Calcolo dell'Indice di Correlazione:** Il dataset filtrato è stato raggruppato in base all'indirizzo di destinazione (`Name`). Per ciascun gruppo, è stato calcolato il coefficiente di correlazione tramite la funzione nativa `.corr()`. 
* **Scelte di Visualizzazione (Scatter Plot):** Per rappresentare la correlazione tra questi due dati è stato scelto uno Scatter Plot. È stata predisposta una griglia orizzontale formata da tre grafici affiancati (1 riga, 3 colonne), in modo da isolare il comportamento degli utenti per ciascun tipo di giocata.
* **Gestione dell'Overplotting:** Considerata l'enorme mole di dati, stampare punti opachi avrebbe generato blocchi di colore indistinguibili (*overplotting*). Per risolvere il problema, è stato impostato un livello di trasparenza elevato (`alpha=0.3`) unito a un bordo sottilissimo. Questa accortezza permette di visualizzare la reale "densità" dei dati: le aree visivamente più marcate indicano la sovrapposizione di migliaia di scommesse con valori identici, rivelando i cluster comportamentali più frequenti.
* **Iniezione Dinamica dei Dati:** Il valore esatto della correlazione matematica (arrotondato a tre cifre decimali) è stato inserito all'interno del titolo di ogni singolo asse per facilitare il confronto immediato tra l'impatto visivo e il riscontro analitico reale.

```python
def plot_fee_amount_correlation(group_data, ax, name, color):
    """ Disegna lo scatterplot tra fee e amount per un singolo gruppo/indirizzo, calcolando e mostrando la correlazione."""
    # calcolo della correlazione tra amount e fee
    corr_val = group_data['amount'].corr(group_data['fee'])
    
    # creazione dello scatterplot
    sea.scatterplot(data=group_data, x='amount', y='fee', ax=ax, alpha=0.3,color=color,edgecolor='white',linewidth=0.3)
    
    # formattazione estetica
    ax.set_title(f'{name}\nCorrelazione: {corr_val:.3f}', fontsize=13, fontweight='bold')
    ax.set_xlabel('Amount (BTC)', fontsize=12)
    ax.set_ylabel('Fee', fontsize=12)
    ax.grid(axis='both', linestyle='--', alpha=0.5)
```

```python
# inizializzazione della figura (1 riga, 3 colonne)
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle('Correlazione tra Fee e Amount per i Top 3 Indirizzi', fontsize=16, fontweight='bold', y=1.08)

scatter_colors = ['#F23645', '#1E3A8A', '#0ECB81'] 

# 3. Iteriamo sui raggruppamenti e iniettiamo assi e colori nella funzione
for idx, (name, group) in enumerate(top3_temporal_data.groupby('Name')):
    plot_fee_amount_correlation(group, axes[idx], name, scatter_colors[idx])

plt.tight_layout()
plt.show()
```

<h2 style="color: #1E3A8A; font-size: 24px; border-bottom: 2px solid #eaeaea; padding-bottom: 5px; margin-top: 30px;">
    7. Analisi Dettagliata sui Top 3 Indirizzi: Intervallo tra Bet Consecutive (Scelte Tecniche)
</h2>

L'ultimo passaggio dell'analisi dettagliata sui tre indirizzi più popolari consiste nell'esaminare la distribuzione del tempo intercorso tra una scommessa e la successiva. Q

Per calcolare questi intervalli e preparare le visualizzazioni, il codice è stato strutturato attraverso le seguenti operazioni:

* **Ordinamento e Shift dei Dati:** Per poter confrontare due eventi consecutivi, il dataframe è stato preliminarmente ordinato in modo rigoroso per indirizzo di destinazione (`Name`) e per data di esecuzione (`datetime`). Successivamente, raggruppando per indirizzo, è stata utilizzata la funzione `.shift(1)` per far "slittare" la colonna del tempo temporale di una riga verso il basso. Questa tecnica permette di allineare sulla stessa riga sia il timestamp della scommessa attuale sia quello della scommessa immediatamente precedente.
* **Calcolo Esplicito dell'Intervallo:** È stata calcolata la differenza algebrica tra le due date. Il risultato transitorio è stato prima convertito in secondi totali e poi diviso per 60, ottenendo una misurazione continua standardizzata in **minuti** (`time_diff_minutes`).
* **Pulizia e Ottimizzazione:** Le colonne di appoggio generate per il calcolo (`prev_datetime` e `time_diff`) sono state rimosse per preservare l'efficienza della memoria prima di invocare il rendering grafico.
* **Scelte di Visualizzazione (Istogramma Logaritmico):** Si è optato per una griglia verticale di tre istogrammi (uno per ciascun indirizzo top 3). Poiché ci si aspetta una distribuzione con una fortissima asimmetria positiva , l'asse Y è stato impostato su una **scala logaritmica**. In assenza di questa scala, le frequenze minori sarebbero state del tutto schiacciate e illeggibili.
* **Filtraggio e Formattazione dell'Asse X:** Per evitare che *outlier* con ore o giorni di distanza distorcessero la finestra di osservazione, il grafico è stato limitato a un tetto massimo di 90 minuti (`limit_minutes=90`).

```python

def prepare_interbet_data(df):
    """
    Ordina i dati e calcola il tempo trascorso (in minuti) tra scommesse consecutive.
    Ritorna un nuovo dataframe con aggiunta la colonna 'time_diff_minutes'.
    """
    # ordino i dati per Nome e Data 
    df_sorted = df.sort_values(by=['Name', 'datetime']).copy()
    
    # recupero la data della scommessa precedente (slittando le righe in basso di 1)
    df_sorted['prev_datetime'] = df_sorted.groupby('Name')['datetime'].shift(1)
    
    # calcolo la differenza temporale esplicita 
    df_sorted['time_diff'] = df_sorted['datetime'] - df_sorted['prev_datetime']
    
    # convertoi la differenza in secondi totali e divido per 60 per ottenere i minuti
    df_sorted['time_diff_minutes'] = df_sorted['time_diff'].dt.total_seconds() / 60.0
    
    # elimino le colonne di appoggio usate per i calcoli per mantenere il dataframe pulito
    df_sorted = df_sorted.drop(columns=['prev_datetime', 'time_diff'])
    
    return df_sorted

    
def plot_interbet_intervals(group_data, ax, name, color, limit_minutes):
    """
    Disegna l'istogramma degli intervalli di tempo tra bet consecutive per un singolo indirizzo,
    utilizzando una scala logaritmica sull'asse Y in modo sicuro. """
    
    # filtro i dati per il limite richiesto di tempo richiesto
    filtered_data = group_data[group_data['time_diff_minutes'] <= limit_minutes]
    
    # creazione dell'istogramma
    sea.histplot(data=filtered_data, x='time_diff_minutes',bins=limit_minutes, color=color,ax=ax,edgecolor='black',  linewidth=0.45)
    
    # impostazione asse Y in scala logaritmica 
    ax.set_yscale('log')
    
    # formattazione estetica
    ax.set_title(f'{name} Inter-bet Intervals', fontsize=13, fontweight='bold')
    ax.set_ylabel('Count (Log Scale)', fontsize=11)
    ax.set_xlim(0, limit_minutes)
    ax.grid(axis='y', linestyle='--', alpha=0.5)
    ax.set_xlabel("Minutes between consecutive bets", fontsize=10)
    ax.set_xticks(range(0, limit_minutes + 1, 5))
```

```python
top3_sorted = prepare_interbet_data(top3_temporal_data)

# configurazione della figura (3 righe, 1 colonna)
fig, axes = plt.subplots(3, 1, figsize=(8, 10))
fig.suptitle("Intervallo tra Bet Consecutive (Scala Y Logaritmica)", fontsize=16, fontweight='bold')

hist_colors = ['#F23645', '#1E3A8A', '#0ECB81'] 

# itero sui raggruppamenti dei 3 indirizzi e disegna i grafici
for idx, (name, group) in enumerate(top3_sorted.groupby('Name')):
    plot_interbet_intervals(group, axes[idx], name, hist_colors[idx], limit_minutes=90)

# formattazione dell'asse X (solo sull'ultimo asse, grazie a sharex=True)
for ax in axes:
    ax.set_xlabel("Minuti tra Bet Consecutive", fontsize=10)

plt.tight_layout()
plt.subplots_adjust(top=0.88)
plt.show()
```

<h2 style="color: #1E3A8A; font-size: 24px; border-bottom: 2px solid #eaeaea; padding-bottom: 5px; margin-top: 30px;">
    8. Analisi delle sequenze di bet (Scelte Tecniche)
</h2>

Il quinto obiettivo del progetto introduce la *Network Analysis* per studiare le sequenze di scommesse (catene), focalizzandosi sull'indirizzo più popolare . Per garantire che le catene siano lineari e prive di ambiguità, l'analisi è stata ristretta al sottoinsieme delle  **"simple bet"**.

Il processo di estrazione e la modellazione del grafo sono stati implementati seguendo questi passaggi algoritmici:

* **Filtraggio Strutturale (Isolamento delle Simple Bet):** Per prima cosa, sono state calcolate le frequenze di input e output per l'intero dataset. sono state mantenute esclusivamente le transazioni caratterizzate da esattamente 1 input e 2 output.
* **Identificazione dell'Output di Resto (Change Address):** Delle transazioni così filtrate, sono state estrapolate solo quelle in cui uno dei due output era destinato all'indirizzo target di Satoshi Dice. il secondo output di queste transazioni è stato classificato come **indirizzo di resto**.
* **Tracciamento dei Collegamenti Logici (Generazione degli Archi):** Per trovare i collegamenti consecutivi, la tabella degli output di resto è stata unita in *inner join* con la tabella `inputs` (effettuando il matching su `txId == prevTxId` e `position == prevTxpos`). Il risultato è stato ulteriormente filtrato per assicurarsi che anche la transazione di destinazione rientrasse nella categoria *simple bet*.
* **Costruzione del Grafo Diretto:** Utilizzando la libreria **NetworkX**, è stato inizializzato un grafo diretto $G=(V,E)$. Ogni identificativo di transazione (`txId`) è stato aggiunto come nodo $V$, mentre i collegamenti trovati nel passaggio precedente sono diventati gli archi orientati $E$. L'ID dell'indirizzo di resto è stato salvato come attributo dell'arco, garantendo la tracciabilità richiesta.
* **Analisi dei Cammini (Weakly Connected Components):** L'architettura stessa del sistema (in-degree e out-degree massimi pari a 1) forza il grafo a presentarsi come un insieme di cammini disgiunti. Sfruttando l'algoritmo di individuazione delle *weakly connected components* di NetworkX, sono state isolate le singole catene, conteggiando il numero di nodi (transazioni) per ciascuna di esse.
* **Visualizzazione Grafica:** La distribuzione delle lunghezze presenta un'asimmetria estrema. Per rappresentarla in modo esaustivo, si è optato per un duplice approccio grafico. A sinistra è stato tracciato un istogramma su scala lineare (limitato visivamente per non schiacciare i dati) per analizzare la concentrazione della massa principale. A destra, l'intero dataset è stato mappato su uno scatterplot in scala bi-logaritmica (Log-Log), strumento statistico per lo studio delle reti complesse al fine di verificare la presenza di distribuzioni a coda lunga

```python

def build_bet_chains_graph(inputs, outputs, bet_outputs, sorted_num_bets):
    """
    Filtra le simple bet, trova i collegamenti tramite gli indirizzi di resto ,costruisce il grafo orientato delle catene di scommesse. e anche le lunghezze delle catene """
    
    #recupero l'id dell'indirizzo più popolare
    target_address_id = sorted_num_bets.tail(1)['addressId'].iloc[0]
    
    # prendo gli id delle transazioni che hanno puntato all'indirizzo più popolare
    target_txs = bet_outputs[bet_outputs['addressId'] == target_address_id]['txId'].unique()
    
    #filtro le tabelle originali per guardare solo le transazioni che hanno puntato sull'indirizzo più popolare
    tx_inputs = inputs[inputs['txId'].isin(target_txs)]
    tx_outputs = outputs[outputs['txId'].isin(target_txs)]
    
    # dataframes con conteggio di quanti input ed output ci sono per ogni transazione
    in_counts = tx_inputs.groupby('txId').agg(count_in=('txId','count'))
    out_counts = tx_outputs.groupby('txId').agg(count_out=('txId','count'))
    
    # unisco le due tabelle dei conteggi usando l'indice (txId)
    merged_counts = in_counts.merge(out_counts, on ='txId')
    
    # condizione selle simple bet ovvero bet con 1 input ed 2 outputs
    condition_simple_bet= (merged_counts['count_in'] == 1) & (merged_counts['count_out'] == 2)
    
    # prendo i'txId'cioe gli index delle righe che rispettano le condizioni delle simple bet
    simple_bet_tx_ids = merged_counts[condition_simple_bet].index
    
    # tengo solo gli outputs che fanno parte delle simple bet
    valid_outputs = tx_outputs[tx_outputs['txId'].isin(simple_bet_tx_ids)]
    
    # visto che so che un indizzo va verso satoshidice, l'altro (diverso da satoshidice) è il resto 
    resto_outputs = valid_outputs[valid_outputs['addressId'] != target_address_id].copy()
    
    #join tra gli output di resto e la tabella inputs per vedere dove vengono spesi
    edges_df = resto_outputs.merge(inputs, left_on=['txId', 'position'], right_on=['prevTxId', 'prevTxpos'], suffixes=('_source', '_dest'))
    
    # verifico che anche la transazione di destinazione sia una simple Bet
    edges_df = edges_df[edges_df['txId_dest'].isin(simple_bet_tx_ids)]

    #costruzione del grafo diretto
    G = nx.DiGraph()
    
    # aggiungo i nodi 
    G.add_nodes_from(simple_bet_tx_ids)
    
    # aggiungo gli archi 
    for _, row in edges_df.iterrows():
        G.add_edge(row['txId_source'], row['txId_dest'], address=row['addressId'])
        
    # prendo tutte le catene (componenti debolmente connesse ) del grafo 
    components = list(nx.weakly_connected_components(G))
    
    # calcola la lunghezza (numero di nodi) per ogni catena, escludendo le catene di lunghezza 1 (scommesse singole che non hanno originato catene) 
    chain_lengths = [len(c) for c in components if len(c) > 1]
    
    return G,chain_lengths

    
```

```python
def plot_hist_distribution(chain_lengths, ax, color):
    """
    Calcola le componenti debolmente connesse (catene) e ne disegna la distribuzione delle lunghezze.
    """
    #configurazione dell'histplot
    sea.histplot(chain_lengths, bins=40,ax=ax,color=color, alpha=0.85,edgecolor='white',linewidth=0.8)
    #applico la scala logaritmica
    ax.set_yscale('log')

    # calcolo della mediana(controllo per evitare errori se non i sono catene
    if chain_lengths:
        mediana = np.median(chain_lengths)
        
        # aggiungo la linea verticale rossa tratteggiata
        ax.axvline(mediana, color='#EF4444', linestyle='--', linewidth=2.5, label=f'Mediana: {mediana:.0f} scommesse')
        
        # mostro la legenda in alto a destra
        ax.legend(fontsize=11, loc='upper right')

    #formattazione 
    ax.set_title('Distribuzione della Lunghezza delle Catene di Bet', fontsize=14, fontweight='bold')
    ax.set_title('Distribuzione Lineare delle Lunghezze', fontsize=14, fontweight='bold')
    ax.set_ylabel('Frequenza (Scala Logaritmica)', fontsize=12)
    ax.set_xlabel('Lunghezza della Catena (Scala Lineare)', fontsize=12)
    ax.grid(axis='y', linestyle='--', alpha=0.5)


def plot_scatter_distribution(chain_lengths, ax, color):
    """ Calcola le componenti debolmente connesse (catene) e ne disegna la distribuzione delle lunghezze
    usando uno scatter plot su scala Log-Log per evidenziare la Power Law (coda lunga).
    """
    if not chain_lengths:
        return
        
    # conto le frequenze per ogni singola lunghezza di catena
    length_counts = Counter(chain_lengths)
    
    # separo l'asse X (lunghezze ordinate) e l'asse Y (la loro frequenza)
    x = sorted(list(length_counts.keys()))
    y = [length_counts[val] for val in x]

    # configurazione dello scatterplot
    sea.scatterplot(x=x, y=y, ax=ax, color=color, s=40, alpha=0.7, edgecolor='black', linewidth=0.4)
    
    # applico la scala logaritmica a entrambi gli assi (grafco Log-Log)
    ax.set_xscale('log')
    ax.set_yscale('log')

    # calcolo della mediana
    mediana = np.median(chain_lengths)
        
    # aggiungo la linea verticale rossa tratteggiata per la mediana
    ax.axvline(mediana, color='#EF4444', linestyle='--', linewidth=2.5, label=f'Mediana: {mediana:.0f} scommesse')
        
    # mostro la legenda
    ax.legend(fontsize=11, loc='upper right')

    # formattazione titolo
    ax.set_title('Distribuzione Log-Log della Lunghezza delle Catene (Scatter)', fontsize=14, fontweight='bold')
    ax.set_xlabel('Lunghezza della Catena (Scala Logaritmica)', fontsize=12)
    ax.set_ylabel('Frequenza (Scala Logaritmica)', fontsize=12)
    
    # griglia: aggiungo anche una griglia leggera sull'asse x per facilitare la lettura della scala logaritmica
    ax.grid(axis='y', linestyle='--', alpha=0.5)
    ax.grid(axis='x', linestyle=':', alpha=0.3)

    
def add_stats_box_to_plot(chain_lengths, ax):
    """
    Calcola le statistiche sulle lunghezze delle catene e le aggiunge come riquadro informativo, ben allineato e posizionato sotto la legenda.
    """
    if not chain_lengths:
        return

    # calcolo delle statistiche
    totale = len(chain_lengths)
    mediana = np.median(chain_lengths)
    media = np.mean(chain_lengths)
    top_5 = sorted(chain_lengths, reverse=True)[:5]
    top_5_str = ", ".join(map(str, top_5))
    catene_oltre_100 = sum(1 for x in chain_lengths if x > 100)

    # testo formattato 
    stats_text = (
        f"STATISTICHE CATENE\n"
        f"--------------------------------\n"
        f"{'Totale Analizzate:':<20} {totale}\n"
        f"{'Lunghezza Mediana:':<20} {mediana:.0f}\n"
        f"{'Lunghezza Media:':<20} {media:.2f}\n"
        f"{'Top 5 Catene:':<20} {top_5_str}\n"
        f"{'Catene > 100 nodi:':<20} {catene_oltre_100}"
    )

    # configurazioe estetica del riquadro
    props = dict(boxstyle='round,pad=0.8', facecolor='white', alpha=0.95, edgecolor='#CBD5E1')

    # Inserimento nel grafico (y=0.82 per abbassarlo, multialignment='left' per inquadrarlo)
    ax.text(0.97, 0.82, stats_text, transform=ax.transAxes, fontsize=11,
            verticalalignment='top', horizontalalignment='right', multialignment='left',bbox=props, family='monospace', color='#1E3A8A')
```

```python
# costruisco il grafo e ottengo le lunghezze di tutte le catene
bet_graph, chain_lengths = build_bet_chains_graph(inputs, outputs, bet_outputs, sorted_num_bets)

# creo la figura (1 riga, 2 colonne) dando molto respiro
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# istogramma (grafico a sinistra)
plot_hist_distribution(chain_lengths, axes[0], color='#0ECB81')

#  scatterplot Log-Log (grafico a destra)
plot_scatter_distribution(chain_lengths, axes[1], color='#0F766E')

# aggiungiamo il box statistico ad entrambi i grafici
add_stats_box_to_plot(chain_lengths, axes[0])
add_stats_box_to_plot(chain_lengths, axes[1])

plt.tight_layout()
plt.show()
```

<h2 style="color: #1E3A8A; font-size: 24px; border-bottom: 2px solid #eaeaea; padding-bottom: 5px; margin-top: 30px;">
    9. De-anonimizzazione delle Catene tramite Web Scraping (Scelte Tecniche)
</h2>

L'ultima fase dell'analisi di rete consiste nell'estrarre le catene di transazioni più lunghe presenti nel grafo e tentare di de-anonimizzare gli indirizzi coinvolti interrogando il servizio pubblico **WalletExplorer.com**.
L'obiettivo è verificare se queste lunghe sequenze di nodi siano riconducibili a singole entità, per automatizzare questa indagine in modo efficiente e invisibile ai sistemi anti-bot, l'approccio è stato il seguente:

* **Estrazione e mapping degli indirizzi:** Tramite la funzione `nx.weakly_connected_components`, il grafo è stato scomposto nelle sue catene isolate, ordinandole in modo decrescente per selezionare solo le prime `h` (le più lunghe).Successivamente, gli identificativi numerici dei nodi (`addressId`) sono stati incrociati con il dataset originale per recuperare gli hash crittografici reali, escludendo a priori gli indirizzi fittizi, contrassegnati dal prefisso `#`.

* **Configurazioni di scraping:** 
Per interrogare la piattaforma senza incorrere in blocchi, come gli errori HTTP 403 o 429, è stato configurato un bot con **Chrome WebDriver**.
È stata usata la libreria `fake_useragent` per camuffare l'impronta del browser a ogni avvio, unita al flag `--disable-blink-features=AutomationControlled` per nascondere la natura automatizzata dello script.L'opzione `--headless=new` ha garantito l'esecuzione silenziosa in background.

* **Simulazione del comportamento umano:** 
Anziché interrogare direttamente gli URL, lo script naviga sulla homepage del sito, individua dinamicamente la barra di ricerca tramite l'attributo `name="q"`, vi digita l'hash e simula la pressione del tasto `INVIO` (`Keys.RETURN`), replicando l'esatta interazione di un utente reale.

* **Gestione dell'overhead:** 
Poiché uno stesso indirizzo può comparire in molteplici archi all'interno della stessa catena o in catene diverse, è stato implementato un dizionario di cache locale (`wallet_cache`).
Il bot interroga la rete solo se l'hash è del tutto nuovo, riducendo drasticamente il numero di richieste HTTP, accelerando l'esecuzione globale.

```python
def get_top_h_chains_addresses(G, h=5):
    """
    Estrae i dati e gli addressId delle h catene più lunghe del grafo. e ritorna una lista di catane (set) con i rispettivi dati
    """
    # estraggo le catene dal grafo e le metto in una lista, (ogni componende e' un set {id1,id2,ecc...})
    components = list(nx.weakly_connected_components(G))
    
    # ordinamento della lista per lunghezza (numero di nodi) in modo decrescente
    sorted_components = sorted(components, key=len, reverse=True)
    
    # prendo solo le prime h catene
    top_h_components = sorted_components[:h]
    
    chains_data = []
    
    for i, comp in enumerate(top_h_components):
        # crea un sottografo per la singola catena, quindi unisce tutti i nodi con i relativi archi (archi presi dal grafo principale)
        subgraph = G.subgraph(comp)
        
        # utilizzo un set in modo da eliminare i duplicati
        address_ids_in_chain = set()
        #per ogni arco iseriosco i data, cioe le etichette nel set
        for u, v, data in subgraph.edges(data=True):
            address_ids_in_chain.add(data['address'])
            
        # salvo le info relative ad ogni componente
        chains_data.append({
            'chain_id': f"Chain_{i+1}",
            'length_nodes': len(comp),
            'length_edges': subgraph.number_of_edges(),
            'address_ids': list(address_ids_in_chain)
        })
        
    return chains_data


def map_addresses_to_hashes(chains_data, mapping_df):
    """ Aggiunge il campo real_hashes ad ogni catena, ed assegna ad esso una lista di hash(gli indirizzi crypto reali) """
    for chain in chains_data:
        # prondo solo gli hash degli id che sono nella catena
        real_hashes = mapping_df[mapping_df['addressId'].isin(chain['address_ids'])]['hash'].tolist()
        
        # tolgo via gli indirizzi hash fittizi (quelli che iniziano con #)
        real_hashes = [h for h in real_hashes if not str(h).startswith('#')]
        
        # aggiungo i valori hash al dizionario della catena
        chain['real_hashes'] = real_hashes
        
    return chains_data
```

```python
from selenium import webdriver
from fake_useragent import UserAgent
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.keys import Keys
import time
from collections import Counter

def get_driver():
    try:
        ua = UserAgent()
        user_agent = ua.random
    except:
        user_agent = "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/120 Safari/537.36"
    options = webdriver.ChromeOptions()
    options.add_argument(f"user-agent={user_agent}")
    options.add_argument("--disable-blink-features=AutomationControlled") #nsaconde la variabile interna 
    options.add_argument("--start-maximized")
    options.add_argument('--headless=new')
    driver = webdriver.Chrome(options=options)
    return driver

def scrape_wallet_explorer(bitcoin_hash, driver, wait):
    """ Usa la barra di ricerca della topbar per cercare l'hash."""
    driver.get("https://www.walletexplorer.com/")

    try:
        # trovo la barra di ricerca in alto a destra (name="q")
        search_bar = wait.until(EC.presence_of_element_located((By.NAME, "q")))
        
        # pulisco la barra e scrivo l'hash
        search_bar.clear()
        search_bar.send_keys(bitcoin_hash)
        
        # premo il tasto invio della tastiera per far partire la ricerca
        search_bar.send_keys(Keys.RETURN)
        
        # prendo l'id del wallet dalla pagina (wallet_name)
        result_element = wait.until(EC.presence_of_element_located((By.CLASS_NAME, "wallet_name")))
        
        # restituisce il testo trovato
        return result_element.text.strip()
        
    except Exception as e:
        print(f"Errore durante la ricerca di {bitcoin_hash}: {e}")
        return "Wallet_Non_Trovato"

def analyze_chains_on_explorer(chains_data, driver, wait):
    """
    Scorre le catene, chiama lo scraper gestendo la cache, e calcola le metriche (wallet predominante, percentuali, ecc.).
    Restituisce una lista di dizionari pronta per pandas.
    """
    wallet_cache = {}
    final_results = []
    
    print("Inizia lo scraping delle catene")
    
    for chain in chains_data:
        chain_id = chain['chain_id']
        hashes_to_find = chain['real_hashes']
        
        print(f"\nAnalisi {chain_id} - ({len(hashes_to_find)} indirizzi validi)")
        
        found_wallets = []
        
        for h in hashes_to_find:
            # gestico il caso in cui il wallet è già stato incontrato
            if h in wallet_cache:
                wallet_id = wallet_cache[h]
            else:
                # altrimenti chiamo lo scraper
                wallet_id = scrape_wallet_explorer(h, driver, wait)
                wallet_cache[h] = wallet_id
                
            
            #inserisco l'id del wallet nei wallet trovati    
            found_wallets.append(wallet_id)
            
        # calcola le metriche 
        if len(found_wallets) > 0:
            wallet_counts = Counter(found_wallets)
            main_wallet, max_occurrences = wallet_counts.most_common(1)[0]
            num_distinct = len(wallet_counts)
            percentage = (max_occurrences / len(found_wallets)) * 100
        else:
            main_wallet, num_distinct, percentage = "Nessuno", 0, 0.0
            
        #salva i risultati
        final_results.append({
            "ID Catena": chain_id,
            "Lunghezza Catena": chain['length_edges'],
            "Tot Indirizzi Analizzati": len(hashes_to_find),
            "Wallet Distinti": num_distinct,
            "% Wallet Predominante": f"{percentage:.2f}%",
            "ID Wallet Predominante": main_wallet
        })
        
    return final_results
```

```python

# prendo le top h catene 
chains_data = get_top_h_chains_addresses(bet_graph, h=10)
# aggiungo ad ogni catena il valori hash corrispondenti agli ids
chains_data = map_addresses_to_hashes(chains_data, mapping)

# avvio Selenium 
driver = get_driver()
wait = WebDriverWait(driver, 10)

try:
    dati_punto_6 = analyze_chains_on_explorer(chains_data, driver, wait)
finally:
    driver.quit()

# creazione e visualizzazione del DataFrame
df_risultati_wallets = pd.DataFrame(dati_punto_6)

print("\n Scraping Completato! Ecco la tabella finale:")
display(df_risultati_wallets)
```

L'esame della tabella dei risultati, generata sui dati estratti dalle 10 catene più lunghe del grafo, offre uno spaccato fondamentale sulle dinamiche di aggregazione dei fondi e sul comportamento delle entità nella rete Bitcoin. Dalla lettura dei dati emergono le seguenti evidenze analitiche:

* **L'eccezione dell'entità singola (Chain_2):** La seconda catena più lunga del dataset spicca per la sua totale centralizzazione. Su ben 441 indirizzi analizzati, è stato rilevato un solo wallet distinto (`[00002c4379]`), portando la percentuale di predominanza al **100.00%**. Questa topologia è il caso di un'unica macro-entità che muove i fondi internamente senza mai interagire con utenti esterni.

* **Predominanza quasi-totale in catene frammentate:** Anche in catene composte da più wallet, l'analisi rivela che queste sono governate di fatto da un singolo attore. Il caso più evidente è la `Chain_5`, dove il wallet predominante controlla il **99.63%** dei 272 nodi, o la `Chain_9` (93.97% di predominanza).

* **Ricorrenza di entità dominanti :** Osservando gli ID dei wallet, emerge come singole entità monopolizzino più catene contemporaneamente. Il wallet `[0025d633f8]` domina sia la catena più lunga (`Chain_1`) sia la terza (`Chain_3`); similmente, il wallet `[0012ea9acf]` controlla sia la `Chain_9` che la `Chain_10`. I picchi di attività sulla rete non sono quindi casuali, ma derivano dalle operazioni continue di pochissime entità.

* **Rapporto inverso tra complessità e predominanza:** La catena più lunga in assoluto (`Chain_1`, 716 indirizzi) presenta paradossalmente il tasso di predominanza più basso della matrice (50.14%) e la massima dispersione (76 wallet distinti). Catene più corte mostrano invece un forte accentramento. Ciò dimostra che quando una sequenza di transazioni si allunga oltre una certa soglia, si interseca inevitabilmente con l'economia reale e con servizi di terze parti, frammentando la tracciabilità.
